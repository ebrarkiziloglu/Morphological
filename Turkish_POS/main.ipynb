# Load Data:

```python
from datasets import load_dataset
# Load only 50 samples  
dataset = load_dataset("hasankursun/turkish-corpus-100b", "pretrain", split="train", streaming=True)
dataset[0]

# pos_testset = []
sample = list(dataset.take(50))
print(sample[0])
```

```python
# pos_testset = [e['text'] for e in sample]
pos_testset = [e['text'].split('.\n', 1)[0] for e in sample]
pos_testset = [e.split('\n', 1)[-1].split('. ', 1)[0] + '.' for e in pos_testset]
pos_testset

for i in range(len(pos_testset)):
    print(f"{i}: {pos_testset[i]}")

keep_indices = [0, 2, 4, 5, 6, 12, 13, 14, 15, 17, 18, 20, 25, 26, 29, 31, 33, 35, 40, 47]
pos_testset = [pos_testset[i] for i in keep_indices]
pos_testset = [s.replace(";", ",") for s in pos_testset]

print('-'*50)
print(pos_testset)

with open('pos_testset.txt', 'w') as f:
    for line in pos_testset:
        f.write(line + '\n')
```

```python
## Load Ground Truth values:
gt = []
with open('pos_testset_groundtruth.csv', 'r') as f:
    for line in f:
        gt.append(line.split(';')[2].strip())

ground_truth = gt[1:]
ground_truth
```

```python
# Eval tool 1 - TULAP POS Tagger
# Same span-aligned eval as VNLP. Convert TULAP's {tokens, tag_list}
# dict into (token, tag) pairs first.

from eval_align import load_gold, evaluate_predictions

gold_rows = load_gold("pos_testset_groundtruth.csv")

start_time = time.time()
tulap_preds = []
for row in gold_rows:
    tagged = tulap_tagger.pos_tag(row["text"])
    tulap_preds.append(list(zip(tagged["tokens"], tagged["tag_list"])))
end_time = time.time()

tulap_scores = evaluate_predictions(gold_rows, tulap_preds)
print(f"Time taken: {end_time - start_time:.4f} seconds")
```
