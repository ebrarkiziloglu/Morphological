```python
from collections import Counter

# Count number of Dative (`Case=Dat`) tokens per sentence for each split
# and aggregate how many sentences have 1, 2, ..., 10 Dative tokens.

# `num_sentences``: total sentences in the split
# `any_dative`: how many sentences contain at least one Dative
# `summary_1_to_10`: dict with counts for sentences having exactly 1–10 Dative tokens
# `full_counter`: full distribution of Dative counts (if you want beyond 10).

def count_dative_per_sentence(sentences):
    counts = []  # number of Dative tokens per sentence
    for sent in sentences:
        tokens = sent.get("tokens", [])
        dat_count = 0
        for tok in tokens:
            feats = tok.get("feats", {}) or {}
            if feats.get("Case") == "Dat":
                dat_count += 1
        counts.append(dat_count)
    return counts

def summarize_counts(counts, max_k=10):
    c = Counter(counts)
    summary = {k: c.get(k, 0) for k in range(1, max_k + 1)}
    any_dative = sum(v for k, v in c.items() if k > 0)
    return c, summary, any_dative
```

```python
# now randomly sample from the train split, I want to sample 5 sentences with 1 Dative token, 5 sentences with 2 Dative tokens, etc.
# and then do the same for 2, 3, 4, 5 Dative tokens.
# and then concatenate them and save as a new file.

# first, sample 5 sentences with 1 Dative token
import random

# Precompute Dative counts for train split if not already done
train_sentences = ud_dataset["train"]
train_counts = count_dative_per_sentence(train_sentences)

# Map: k -> list of indices of sentences with exactly k Dative tokens
indices_by_k = {}
for idx, c in enumerate(train_counts):
    if c == 0:
        continue
    indices_by_k.setdefault(c, []).append(idx)

# For k = 1..5, sample up to 5 sentences each (without replacement)
sampled_indices_by_k = {}
for k in range(1, 6):
    candidates = indices_by_k.get(k, [])
    if not candidates:
        sampled_indices_by_k[k] = []
        continue
    n_sample = min(5, len(candidates))
    sampled_indices_by_k[k] = random.sample(candidates, n_sample)

# Collect sampled sentences in a single list, keeping track of (k, index)
sampled_train_sentences = []
for k in range(1, 6):
    for idx in sampled_indices_by_k[k]:
        sent = train_sentences[idx]
        sampled_train_sentences.append({
            "k_dative": k,
            "orig_index": idx,
            "sent_id": sent.get("sent_id"),
            "text": sent.get("text"),
            "tokens": sent.get("tokens"),
        })

len(sampled_train_sentences), {k: len(v) for k, v in sampled_indices_by_k.items()}
```

```python
import csv
import os
from collections import defaultdict
from typing import Dict, Set

from morphological_analysis_helpers import load_ud_dataset, UD_GERMAN_GSD_SPLITS
```

```python
ud_base_path = "../data/UD_German-GSD"
print(f"Loading UD German-GSD from: {ud_base_path}")
dataset = load_ud_dataset(ud_base_path, splits_filemap=UD_GERMAN_GSD_SPLITS)
```

```python
dataset['test'][0]
```

```python
# Dictionary: lemma -> { "Acc": set(forms), "Dat": set(forms) }
# Also capturing all other cases just in case, but formatting mostly for Acc and Dat
vocab_cases = defaultdict(lambda: defaultdict(set))

splits = ["train", "dev", "test"]

for split in splits:
    if split not in dataset:
        continue
    for sent in dataset[split]:
        for token in sent.get("tokens", []):
            # print(token)
            # print('=' * 50)
            lemma = token.get("lemma", "")
            form = token.get("form", "")
            upos = token.get("upos", "")
            feats = token.get("feats", {}) or {}
            
            case = feats.get("Case")
            number = feats.get("Number")
            gender = feats.get("Gender")
            
            if lemma and form and case:
                if case in {"Acc", "Dat"}:
                    vocab_cases[lemma][(case, number, upos)].add(form)
```

```python
count = 0

for (lemma, values) in vocab_cases.items():
    count += 1
    print(f'lemma: {lemma}:')
    for (key, value) in values.items():
        print(f'\t{key}: {value}')
    if count > 10:
        break


output_csv = "german_ud_cases_dictionary.csv"

formats = {}

print(f"Writing to {output_csv}...")
with open(output_csv, 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(["Lemma", "Case", "Number", "Upos", "Forms"])
    
    for lemma in sorted(vocab_cases.keys()):
        cases = vocab_cases[lemma]
        for key, forms in cases.items():
            forms_str = " | ".join(sorted(forms))
            writer.writerow([lemma, key[0], key[1], key[2], forms_str])

print(f"Done! Saved {len(vocab_cases)} entries to {output_csv}")

# for key in formats:
#     print(f'{key}: {formats[key]}')
```
